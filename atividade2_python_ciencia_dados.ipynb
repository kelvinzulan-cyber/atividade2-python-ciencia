Este notebook apresenta a análise das características financeiras do dataset `default_of_credit_card_clients`.


## Importação das bibliotecas e leitura da planilha

Nesta etapa são importadas as bibliotecas necessárias e a planilha é carregada para análise.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Leitura da planilha
df = pd.read_excel('default_of_credit_card_clients.xls', header=1)

# Visualizando as primeiras linhas
df.head()
```

## Exercício 1

Criação de listas com os nomes das características financeiras restantes.


```python
bill_feats = ['BILL_AMT1', 'BILL_AMT2', 'BILL_AMT3', 'BILL_AMT4', 'BILL_AMT5', 'BILL_AMT6']
pay_feats = ['PAY_AMT1', 'PAY_AMT2', 'PAY_AMT3', 'PAY_AMT4', 'PAY_AMT5', 'PAY_AMT6']

financial_feats = bill_feats + pay_feats

financial_feats
```

## Exercício 2

Uso do `.describe()` para examinar as estatísticas dos valores da fatura.


```python
df[bill_feats].describe()
```

**Comentário:** É possível observar valores mínimos, máximos, média e desvio padrão das faturas. Isso faz sentido, pois os clientes possuem diferentes limites, gastos e comportamentos financeiros.


## Exercício 3

Visualização das características de valor da fatura usando histogramas em uma grade 2 por 3.


```python
df[bill_feats].hist(bins=20, layout=(2, 3), figsize=(12, 8))
plt.tight_layout()
plt.show()
```

## Exercício 4

Resumo estatístico das características de valor do pagamento.


```python
df[pay_feats].describe()
```

**Comentário:** O resultado faz sentido, pois alguns clientes pagam valores altos, enquanto outros pagam pouco ou até não realizam pagamento em alguns meses.


## Exercício 5

Histogramas das características de pagamento da fatura, com rotação dos rótulos do eixo x.


```python
df[pay_feats].hist(bins=20, layout=(2, 3), figsize=(12, 8), xrot=45)
plt.tight_layout()
plt.show()
```

## Exercício 6

Uso de uma máscara booleana para contar quantos valores de pagamento são iguais a zero.


```python
pay_zero_mask = df[pay_feats] == 0

pay_zero_mask.sum()
```

**Comentário:** O resultado faz sentido porque nos histogramas anteriores existe uma concentração grande de pagamentos próximos de zero.


## Exercício 7

Transformação logarítmica dos pagamentos diferentes de zero usando `apply()` e `np.log10()`.


```python
pay_non_zero = df[pay_feats].mask(pay_zero_mask)

pay_log = pay_non_zero.apply(np.log10)

pay_log.hist(bins=20, layout=(2, 3), figsize=(12, 8), xrot=45)
plt.tight_layout()
plt.show()
```

**Comentário final:** Ao ignorar os pagamentos iguais a zero e aplicar o logaritmo, a visualização fica melhor distribuída, facilitando a análise dos valores de pagamento.
